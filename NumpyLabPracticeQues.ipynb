```python
# 12. Create an array and demonstrate the difference between a slice/view and a copy.

import numpy as np

a = np.array([1, 2, 3, 4, 5])

view = a[1:4]
copy = a[1:4].copy()

view[0] = 100
copy[1] = 200

print("Original array:", a)
print("View:", view)
print("Copy:", copy)
```

```python
# 16. Create a 5 × 5 matrix containing 1–25 and extract the first row, last column, middle element,
#  and top-left 3 × 3 matrix.

import numpy as np

a = np.arange(1, 26).reshape(5, 5)

print("Matrix:")
print(a)

print("First row:", a[0])
print("Last column:", a[:, -1])
print("Middle element:", a[2, 2])
print("Top-left 3x3:")
print(a[:3, :3])
```

```python
# 23. Find the transpose of a matrix and calculate its dot product with the original matrix.

import numpy as np

a = np.array([
    [1, 2],
    [3, 4]
])

t = a.T

print("Transpose:")
print(t)

print("Dot product:")
print(np.dot(t, a))
```

```python
# 28. Given marks = np.array([35,72,48,90,55,28]), replace marks below 40 with 0 using
# np.where().

import numpy as np

marks = np.array([35, 72, 48, 90, 55, 28])

result = np.where(marks < 40, 0, marks)

print(result)
```

```python
# 29. Given an income array, classify observations as &#39;High&#39; if income ≥ 50,000 and &#39;Low&#39;
# otherwise.

import numpy as np

income = np.array([30000, 55000, 45000, 80000, 60000])

result = np.where(income >= 50000, "High", "Low")

print(result)
```

```python
# 30. Replace all values greater than 30 in an array with 100 using np.where().

import numpy as np

a = np.array([10, 25, 35, 40, 20, 50])

result = np.where(a > 30, 100, a)

print(result)
```

```python
# 31. Given a 3 × 3 matrix, calculate its total sum, mean, variance, and standard deviation.

import numpy as np

a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Variance:", np.var(a))
print("Standard deviation:", np.std(a))
```
