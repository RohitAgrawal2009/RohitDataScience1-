```python
def inc_val(val):
  val = val + 1 # This 'val' is local to the function

x = 7
inc_val(x)
print(x)
# Integers are immutable in Python. When 'x' (an integer) is passed to 'inc_val', a new local variable 'val' is created inside the function and assigned the value of 'x'.
# Changes to 'val' inside the function do not affect the original 'x' in the global scope.
```

```python
#Q7(e)
val1 = int(input("Enter First Number: "))
val2 = int(input("Enter Second Number: "))

def swap_message(a, b):
    print("Before Swap")
    print(a)
    print(b)
    temp = a
    a = b
    b = temp
    print("After Swap")
    print(a)
    print(b)

swap_message(val1, val2)
```

```python
#Q8(d)
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

# Test the function
print(f"Before: {counter}")
increment_counter()
print(f"After: {counter}")
```
