36. Matrix transformation and analysis: Construct a 5×5 matrix containing integers 21 to 45. Extract the central 3×3 matrix, transpose it, and calculate the matrix product of the extracted matrix with its transpose. Calculate row-wise sums of the product and identify its maximum and all positions tied at that maximum. Explain dimensions at every stage.

```python
# Initial 5x5 matrix -> Shape: (5, 5)
mat5x5 = np.arange(21, 46).reshape(5, 5)

# Central 3x3 matrix -> Shape: (3, 3)
central = mat5x5[1:4, 1:4]

# Transpose -> Shape: (3, 3)
central_T = central.T

# Product of (3, 3) and (3, 3) -> Shape: (3, 3)
prod = central @ central_T

# Row-wise sum -> Shape: (3,)
row_sums = prod.sum(axis=1)
max_val = row_sums.max()
max_indices = np.where(row_sums == max_val)[0]

print("5x5 Matrix:\n", mat5x5)
print("Extracted 3x3:\n", central)
print("Product Matrix:\n", prod)
print("Row-wise sums:", row_sums)
print("Maximum sum:", max_val)
print("Indices tied at maximum:", max_indices.tolist())
```

37. Sales and revenue analysis: Create a 6×4 matrix of quantities sold over six days for four products. Use unit prices [150, 240, 90, 320]. Calculate daily revenue, total revenue per product, the highest-revenue day, and the best-selling product by quantity. Reduce prices by 10% for products whose total quantity exceeds 100, then recalculate total revenue without modifying original prices.

```python
np.random.seed(42)
quantities = np.random.randint(10, 30, size=(6, 4))
prices = np.array([150, 240, 90, 320])

daily_rev = quantities @ prices
prod_qty = quantities.sum(axis=0)
prod_rev = prod_qty * prices

highest_day = np.argmax(daily_rev) + 1
best_product = np.argmax(prod_qty) + 1

discounted_prices = np.where(prod_qty > 100, prices * 0.9, prices.astype(float))
recalculated_total = (quantities @ discounted_prices).sum()

print("Quantities:\n", quantities)
print("Daily Revenue:", daily_rev)
print("Total Revenue per Product:", prod_rev)
print(f"Highest-revenue day: Day {highest_day}")
print(f"Best-selling product: Product {best_product}")
print("Total quantity per product:", prod_qty)
print("Recalculated Total Revenue (with 10% discount):", recalculated_total)
print("Original prices unmodified:", prices)
```

38. Random data cleaning and statistics: Generate 100 normal observations with mean 50 and standard deviation 12 using a fixed seed. Count values outside 30-70 and replace them with the nearest boundary using np.where(). Compare original and cleaned mean, variance, standard deviation, minimum, and maximum. Explain the statistical effects of boundary replacement.

```python
np.random.seed(0)
data = np.random.normal(loc=50, scale=12, size=100)

outside_mask = (data < 30) | (data > 70)
outside_count = np.sum(outside_mask)

# Boundary replacement (winsorization)
cleaned = np.where(data < 30, 30, np.where(data > 70, 70, data))

print("Count outside 30-70:", outside_count)
print(
    f"Original -> Mean: {data.mean():.2f}, Var: {data.var():.2f}, Std: {data.std():.2f}, Min: {data.min():.2f}, Max: {data.max():.2f}"
)
print(
    f"Cleaned  -> Mean: {cleaned.mean():.2f}, Var: {cleaned.var():.2f}, Std: {cleaned.std():.2f}, Min: {cleaned.min():.2f}, Max: {cleaned.max():.2f}"
)
# Boundary replacement compresses extreme tails into single boundary points,
# keeping sample size intact while reducing overall variance and standard deviation.
```

39. Multi-branch financial reconciliation: Combine the three earnings Series by ordinary addition. Explain missing values, identify quarters common to all branches, and calculate their combined earnings. Construct a separate result treating absent branch records as zero and explain how its interpretation differs.
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})   branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})   branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})

```python
branch_a = pd.Series({"Q1": 120, "Q2": 145, "Q4": 180})
branch_b = pd.Series({"Q2": 110, "Q3": 135, "Q4": 155})
branch_c = pd.Series({"Q1": 95, "Q3": 125, "Q4": 140})

# Ordinary addition: Missing values cause NaN because presence in all series is required
ordinary_sum = branch_a + branch_b + branch_c

# Common quarters present in all three
common_quarters = branch_a.index.intersection(
    branch_b.index.intersection(branch_c.index)
)
common_earnings = (branch_a + branch_b + branch_c).loc[common_quarters]

# Zero-filled sum: Treats missing branch records as zero revenue
zero_filled_sum = branch_a.add(branch_b, fill_value=0).add(
    branch_c, fill_value=0
)

print("Ordinary Addition:\n", ordinary_sum)
print("\nCommon Quarters:", common_quarters.tolist())
print("Combined earnings for common quarters:\n", common_earnings)
print("\nResult with absent treated as 0:\n", zero_filled_sum)
```

40. Series data integrity: Preserve the required index order, identify missing student records, and filter valid scores of at least 70. Convert the filtered Series to a dictionary, reconstruct it as a Series, and verify that labelled values match. Explain why missing records should not automatically be zero marks.
scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}   required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']   s = pd.Series(scores, index=required)

```python
scores = {"R101": 78, "R102": 65, "R104": 92, "R105": 48}
required = ["R105", "R103", "R101", "R106", "R104", "R102"]

s = pd.Series(scores, index=required)
missing_students = s[s.isna()].index.tolist()
valid_scores = s[s >= 70]

filtered_dict = valid_scores.to_dict()
reconstructed_s = pd.Series(filtered_dict)
matches = valid_scores.equals(reconstructed_s)

print("Constructed Series:\n", s)
print("\nMissing student records:", missing_students)
print("\nScores >= 70:\n", valid_scores)
print("\nDictionary:", filtered_dict)
print("Reconstructed matches original:", matches)
# Missing records indicate absence/unrecorded data rather than assessed zero performance;
# converting them to 0 would skew class metrics like mean and standard deviation.
```
