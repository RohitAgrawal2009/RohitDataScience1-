## Topic 6: Tuples and Immutability

As seen in Q2.2 (`my_tuple[0] = 4 # tuple is immutable its can not be change`), tuples are immutable. This means you cannot directly add new elements to an existing tuple. However, you can create a *new* tuple by concatenating it with other tuples or by converting it to a list, adding elements, and then converting it back to a tuple.

```python
# Example: Adding strings to a tuple (by creating a new tuple)
my_tuple = (1, 2, 3)
print(f"Original tuple: {my_tuple}")

# Method 1: Concatenate with a new tuple containing strings
new_strings_tuple = ('hello', 'world')
combined_tuple = my_tuple + new_strings_tuple
print(f"New tuple after concatenation: {combined_tuple}")

# Method 2: Convert to a list, add strings, then convert back to tuple
temp_list = list(my_tuple)
temp_list.append('python')
temp_list.append('programming')
tuple_from_list = tuple(temp_list)
print(f"New tuple created via list conversion: {tuple_from_list}")
```

```python
# Q7.1. Write a func on my_abs(val) that returns the absolute value of val without using Python's built-in abs().
# Test it with a posi ve number, a nega ve number, and 0.

def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(7))
print(my_abs(-7))
print(my_abs(0))
```
